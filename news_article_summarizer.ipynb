# News Article Summarizer

A Natural Language Processing (NLP) project that uses a pre-trained Transformer model to automatically generate concise summaries from news articles.

```python
pip install transformers torch sentencepiece
```

```python
from transformers import pipeline
```

### 2. Load the Model

```python
summarizer = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6"
)
```

### 3. Generate Summary

```python
def summarize_article(text):
    """
    Generate a summary for a given news article.
    """

    result = summarizer(
        text,
        max_length=60,
        min_length=15,
        do_sample=False
    )

    return result[0]["summary_text"]
```

### 4. Test with an article

```python
article = """
NASA scientists have announced the discovery of a new planet
outside our solar system. The planet was detected using advanced
space telescopes and is located hundreds of light-years away from
Earth.

Researchers said that the newly discovered planet has attracted
attention because its size and distance from its star may provide
conditions that could support further scientific investigation.

Scientists are now planning additional observations to study the
planet's atmosphere, temperature and chemical composition. The
research team said that more observations will be required before
they can determine whether the planet could support life.
"""

summary = summarize_article(article)

print("Original Article:")
print(article)

print("\nGenerated Summary:")
print(summary)
```

### 5. Calculate text reduction

```python
original_words = len(article.split())
summary_words = len(summary.split())

reduction = ((original_words - summary_words) / original_words) * 100

print("Original words:", original_words)
print("Summary words:", summary_words)
print(f"Text reduction: {reduction:.2f}%")
```

### 6. User Input

```python
news_article = input("Enter your news article:\n")

summary = summarize_article(news_article)

print("\nGenerated Summary:")
print(summary)
```
